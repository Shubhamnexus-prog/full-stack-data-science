```python
from wordcloud import WordCloud, STOPWORDS
import matplotlib.pyplot as plt

text = """
Python is easy. Python is powerful.
The Python language is useful for NLP.
NLP is useful for machine learning.
"""

wordcloud = WordCloud(
    width=1000,
    height=500,
    background_color="white",
    stopwords=STOPWORDS,
    collocations=False
).generate(text)

plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.show()
```

```python

from wordcloud import WordCloud
import matplotlib.pyplot as plt

# Step 1: Sample text
text = """
Python is a powerful programming language.
Python is useful for machine learning.
Machine learning is useful for NLP.
NLP helps computers understand human language.
Python makes NLP projects easier.
"""

# Step 2: Generate the word cloud
wordcloud = WordCloud(
    width=1000,
    height=500,
    background_color="white",
    max_words=100
).generate(text)

# Step 3: Display the result
plt.figure(figsize=(12, 6))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title("NLP Word Cloud")
plt.show()
```

# chunking

```python
import nltk

sentence = "The intelligent boy is playing cricket"

# Tokenization
tokens = nltk.word_tokenize(sentence)

# POS tagging
tagged_words = nltk.pos_tag(tokens)

print("POS Tags:")
print(tagged_words)

# Chunking grammar
grammar = r"""
    NP: {<DT>?<JJ.*>*<NN.*>+}
    VP: {<VB.*>+}
"""

# Create chunk parser
chunk_parser = nltk.RegexpParser(grammar)

# Parse the tagged sentence
result = chunk_parser.parse(tagged_words)

print(result)
result.pretty_print()
```
