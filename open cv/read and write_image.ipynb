```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

image_path = r"E:\full stack data science\open cv\dog.jpg"

# Check file
print("File exists:", os.path.exists(image_path))
print("File size:", os.path.getsize(image_path), "bytes")

# Read image using NumPy
image_data = np.fromfile(image_path, dtype=np.uint8)

# Decode image
img = cv2.imdecode(image_data, cv2.IMREAD_COLOR)

if img is None:
    print("OpenCV cannot decode this image.")
else:
    print("Image loaded successfully!")
    print("Image shape:", img.shape)

    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(8, 6))
    plt.imshow(img_rgb)
    plt.title("Dog Image")
    plt.axis("off")
    plt.show()
```

```python
from PIL import Image

input_path = r"E:\full stack data science\open cv\dog.jpg"
output_path = r"E:\full stack data science\open cv\dog_fixed.jpg"

img = Image.open(input_path)

# Convert to RGB
img = img.convert("RGB")

# Save as a new JPEG
img.save(output_path, format="JPEG", quality=95)

print("New image saved:", output_path)
```
